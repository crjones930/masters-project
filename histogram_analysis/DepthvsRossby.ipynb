### Rossby number

As a bunch of the simulations have 0 rotation, I actually want to calculate the inverse of the Rossby number.
$$\frac{1}{\text{Ro}} = \frac{\tau_\text{conv}}{P_0} = \frac{ \Omega_0\tau_\text{conv}}{2\pi}$$

```python
Rotations = np.array([5,1,0,0,1,5,0])
t_convs = np.array([1898298.8378527947,1988021.1538461538,2287938.1320642475,
           13042535.967619125,10125661.234048655,22260052.264808364,37317109634.5515])

sol_rot = 2.702e-6

inv_rossby = sol_rot * Rotations * t_convs / (2 * np.pi)
inv_rossby
```

```python
inv_rossby_CS = inv_rossby[:3]
inv_rossby_YS = inv_rossby[3:6]
inv_rossby_YS_3D = inv_rossby[-1]
```
