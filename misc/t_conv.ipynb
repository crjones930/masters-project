```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
t_conv = 2.3e6 # 644050.7499677864 # rough est from another script...
Omega_Sun = 2.702e-6
```

```python
inv_Ro = np.linspace(0.5, 10)
```

```python
def Omega(inv_Ro, t_conv):
    return 2 * np.pi * inv_Ro / t_conv
```

```python
# Ro = P/t_conv = 2pi/\Omega / t_conv
# Omega = 2pi / Ro / t_conv = 2pi inv_Ro / t_conv

# plt.plot(inv_Ro, Omega(inv_Ro, 2.3e6) / Omega_Sun, label = 'Sun', c='tab:olive')
plt.plot(inv_Ro, Omega(inv_Ro, 644050.7499677864) / Omega_Sun, label = '20M', c='k')
plt.plot(inv_Ro, Omega(inv_Ro, 1.1e6) / Omega_Sun, label = '20M (Baraffe 2023)', c='tab:blue')
plt.plot(inv_Ro, Omega(inv_Ro, 1.2e6) / Omega_Sun, label = '20M (Morton 2025)', c='tab:blue', ls='--')
plt.plot(inv_Ro, Omega(inv_Ro, 3.54e6) / Omega_Sun, label = '20M (MESA full)', c='tab:red') # calculating t_conv over whole conv. region.
plt.plot(inv_Ro, Omega(inv_Ro, 2.17e6) / Omega_Sun, label = '20M (MESA reduced)', c='tab:red', ls='--') # only calculating t_conv over domain used in MUSIC sims.
plt.xlabel('inverse Rossby [1/Ro]')
plt.ylabel('Omega [sun]')
plt.legend()
plt.grid()
plt.xlim(0.5,10)
# plt.ylim(0.0,18)
print(f'Ro : 0.5 >>> {Omega(2, 644050.7499677864) / Omega_Sun}')
print(f'Ro : 0.1 >>> {Omega(10, 644050.7499677864) / Omega_Sun}')
plt.savefig("Omega_vs_Rossby.png", bbox_inches="tight")
```
